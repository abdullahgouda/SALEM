```python
import numpy as np
import pandas as pd

# =========================================================
# Priority & Resource Allocation Dataset Generator
# SALEM Graduation Project
# =========================================================

np.random.seed(42)

severity_map = {
    "Low": 0.3,
    "Medium": 0.6,
    "High": 0.9
}

teams = ["Team_A", "Team_B", "Team_C"]

data = []

for i in range(1, 12001):

    # =====================================================
    # Incident Severity
    # =====================================================

    severity_label = np.random.choice(
        ["Low", "Medium", "High"],
        p=[0.4, 0.35, 0.25]
    )

    severity_score = severity_map[severity_label]

    # =====================================================
    # Operational Features
    # =====================================================

    area_load = round(np.random.beta(2, 2), 2)

    available_teams = np.random.randint(1, 5)

    team_skill_match = round(np.random.beta(5, 2), 2)

    historical_team_performance = round(
        np.random.beta(4, 2), 2
    )

    citizen_trust_score = round(
        np.random.uniform(0.3, 1.0), 2
    )

    reports_nearby_1h = np.random.poisson(3)

    # =====================================================
    # Priority Score Logic
    # =====================================================

    priority_score = (
    severity_score * 0.55
    + area_load * 0.20
    + (1 - historical_team_performance) * 0.10
    + (1 - citizen_trust_score) * 0.10
    + (reports_nearby_1h / 10) * 0.05
    + np.random.normal(0, 0.05)
)

    # =====================================================
    # Priority Classification
    # =====================================================

    if priority_score >= 0.65:
        priority_level = "P1"

    elif priority_score >= 0.42:
        priority_level = "P2"

    else:
        priority_level = "P3"

    # =====================================================
    # Team Assignment Logic
    # =====================================================

    if severity_label == "High":

        if team_skill_match >= 0.80:
            assigned_team = "Team_A"

        elif historical_team_performance >= 0.70:
            assigned_team = "Team_B"

        else:
            assigned_team = "Team_C"

    elif severity_label == "Medium":

        if historical_team_performance >= 0.75:
            assigned_team = "Team_B"

        else:
            assigned_team = np.random.choice(
                ["Team_A", "Team_C"]
            )

    else:

        assigned_team = np.random.choice(
            ["Team_B", "Team_C"]
        )

    # =====================================================
    # Resource Allocation
    # =====================================================

    if priority_level == "P1":

        suggested_resources = "High"

        resources_count = np.random.randint(4, 7)

    elif priority_level == "P2":

        suggested_resources = "Medium"

        resources_count = np.random.randint(2, 4)

    else:

        suggested_resources = "Low"

        resources_count = 1

    # =====================================================
    # Store Row
    # =====================================================

    data.append({

        "incident_id": i,

        "severity_label": severity_label,

        "severity_score": severity_score,

        "area_load": area_load,

        "available_teams": available_teams,

        "team_skill_match": team_skill_match,

        "historical_team_performance":
            historical_team_performance,

        "citizen_trust_score":
            citizen_trust_score,

        "reports_nearby_1h":
            reports_nearby_1h,

        "priority_score":
            round(priority_score, 3),

        "priority_level":
            priority_level,

        "assigned_team":
            assigned_team,

        "suggested_resources":
            suggested_resources,

        "resources_count":
            resources_count
    })

# =========================================================
# Create DataFrame
# =========================================================

df_priority = pd.DataFrame(data)

# =========================================================
# Save Dataset
# =========================================================

df_priority.to_csv(
    "priority_allocation_dataset_v2.csv",
    index=False
)

# =========================================================
# Dataset Overview
# =========================================================

print("\n✅ Dataset Generated Successfully")
print("=" * 50)

print("\nShape:")
print(df_priority.shape)

print("\nPriority Distribution:")
print(df_priority["priority_level"].value_counts())

print("\nAssigned Team Distribution:")
print(df_priority["assigned_team"].value_counts())

print("\nSample Data:")
print(df_priority.head())
```
