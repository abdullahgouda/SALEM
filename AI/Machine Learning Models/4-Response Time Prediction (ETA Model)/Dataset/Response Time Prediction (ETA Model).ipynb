```python
import numpy as np
import pandas as pd

# =========================================================
# Response Time Prediction Dataset Generator
# SALEM Graduation Project
# =========================================================

np.random.seed(42)

# =========================================================
# Categories
# =========================================================

incident_types = [

    "Electricity",

    "Water",

    "Road",

    "Gas",

    "Waste"
]

area_types = [

    "Residential",

    "Commercial",

    "Industrial"
]

teams = [

    "Team_A",

    "Team_B",

    "Team_C"
]

# =========================================================
# Priority Speed Factors
# Lower = Faster Response
# =========================================================

priority_map = {

    "P1": 0.7,

    "P2": 1.0,

    "P3": 1.3
}

# =========================================================
# Store Data
# =========================================================

data = []

# =========================================================
# Generate Dataset
# =========================================================

for i in range(1, 35001):

    # =====================================================
    # Incident Information
    # =====================================================

    incident_type = np.random.choice(
        incident_types
    )

    severity_score = round(
        np.random.uniform(0.2, 1.0),
        2
    )

    priority = np.random.choice(

        ["P1", "P2", "P3"],

        p=[0.3, 0.4, 0.3]
    )

    assigned_team = np.random.choice(
        teams
    )

    area_type = np.random.choice(
        area_types
    )

    # =====================================================
    # Time Features
    # =====================================================

    hour = np.random.randint(0, 24)

    day_of_week = np.random.randint(0, 7)

    # =====================================================
    # Operational Features
    # =====================================================

    area_load = round(

        np.random.beta(2, 2),

        2
    )

    team_efficiency = round(

        np.random.beta(4, 2),

        2
    )

    traffic_level = round(

        np.random.beta(2, 2),

        2
    )

    # =====================================================
    # Base ETA Logic
    # =====================================================

    base_time = (

        6

        + (1 - severity_score) * 4
    )

    # =====================================================
    # Factors
    # =====================================================

    priority_factor = priority_map[
        priority
    ]

    load_factor = (
        1 + area_load
    )

    efficiency_factor = (

        1 - (
            team_efficiency * 0.5
        )
    )

    traffic_factor = (

        1 + (
            traffic_level * 1.2
        )
    )

    # =====================================================
    # Night Penalty
    # =====================================================

    if hour < 7 or hour > 20:

        night_penalty = 1.2

    else:

        night_penalty = 1.0

    # =====================================================
    # Weekend Penalty
    # =====================================================

    if day_of_week in [5, 6]:

        weekend_penalty = 1.1

    else:

        weekend_penalty = 1.0

    # =====================================================
    # Final ETA Calculation
    # =====================================================

    eta = (

        base_time

        * priority_factor

        * load_factor

        * efficiency_factor

        * traffic_factor

        * night_penalty

        * weekend_penalty

        + np.random.normal(0, 0.8)
    )

    # Avoid Negative Values
    eta = round(
        max(0.5, eta),
        2
    )

    # =====================================================
    # Store Row
    # =====================================================

    data.append({

        "incident_id":
            i,

        "incident_type":
            incident_type,

        "severity_score":
            severity_score,

        "priority_level":
            priority,

        "assigned_team":
            assigned_team,

        "area_type":
            area_type,

        "hour":
            hour,

        "day_of_week":
            day_of_week,

        "area_load":
            area_load,

        "team_efficiency":
            team_efficiency,

        "traffic_level":
            traffic_level,

        "estimated_resolution_time":
            eta
    })

# =========================================================
# Create DataFrame
# =========================================================

df_eta = pd.DataFrame(data)

# =========================================================
# Save Dataset
# =========================================================

df_eta.to_csv(

    "response_time_dataset_v2.csv",

    index=False
)

# =========================================================
# Dataset Overview
# =========================================================

print("\n✅ ETA Dataset Generated Successfully")

print("=" * 50)

print("\nShape:")

print(df_eta.shape)

print("\nETA Statistics:")

print(

    df_eta[
        "estimated_resolution_time"
    ].describe()
)

print("\nPriority Distribution:")

print(

    df_eta[
        "priority_level"
    ].value_counts()
)

print("\nSample Data:")

print(df_eta.head())
```
