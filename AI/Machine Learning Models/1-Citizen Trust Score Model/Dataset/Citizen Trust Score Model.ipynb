```python
import numpy as np
import pandas as pd

np.random.seed(42)

def generate_citizen(profile, citizen_id):

    if profile == "trusted":

        total = np.random.randint(5, 20)

        valid = np.random.randint(
            int(0.75 * total),
            total
        )

        duplicate = np.random.randint(0, 3)

        fake = np.random.choice(
            [0, 1],
            p=[0.95, 0.05]
        )

        recent = np.random.randint(0, 4)

        avg_severity = round(
            np.random.uniform(1.5, 3.0),
            2
        )

        label = np.random.choice(
            [1, 0],
            p=[0.95, 0.05]
        )

    elif profile == "normal":

        total = np.random.randint(10, 35)

        valid = np.random.randint(
            int(0.45 * total),
            int(0.8 * total)
        )

        duplicate = np.random.randint(1, 6)

        fake = np.random.randint(0, 3)

        recent = np.random.randint(2, 8)

        avg_severity = round(
            np.random.uniform(1.0, 4.0),
            2
        )

        label = np.random.choice(
            [1, 0],
            p=[0.75, 0.25]
        )

    else:  # spam

        total = np.random.randint(25, 60)

        valid = np.random.randint(
            0,
            int(0.45 * total)
        )

        duplicate = np.random.randint(
            4,
            int(0.6 * total)
        )

        fake = np.random.randint(1, 10)

        recent = np.random.randint(8, 25)

        avg_severity = round(
            np.random.uniform(2.5, 5.0),
            2
        )

        label = np.random.choice(
            [0, 1],
            p=[0.9, 0.1]
        )

    return {
        "citizen_id": citizen_id,
        "total_reports": total,
        "valid_reports": valid,
        "duplicate_reports": duplicate,
        "fake_image_count": fake,
        "avg_severity_reported": avg_severity,
        "reports_last_30_days": recent,
        "trust_label": label
    }


# ===========================
# Generate Dataset
# ===========================

data = []
citizen_id = 1

for _ in range(6000):
    data.append(generate_citizen("trusted", citizen_id))
    citizen_id += 1

for _ in range(8000):
    data.append(generate_citizen("normal", citizen_id))
    citizen_id += 1

for _ in range(6000):
    data.append(generate_citizen("spam", citizen_id))
    citizen_id += 1


df = pd.DataFrame(data)

# ===========================
# Feature Engineering
# ===========================

df["valid_ratio"] = (
    df["valid_reports"] /
    df["total_reports"]
)

df["duplicate_rate"] = (
    df["duplicate_reports"] /
    df["total_reports"]
)

df["fake_image_ratio"] = (
    df["fake_image_count"] /
    df["total_reports"]
)

# ===========================
# Save
# ===========================

df.to_csv(
    "citizen_trust_dataset_v2.csv",
    index=False
)

print("✅ Dataset Shape:", df.shape)

print(df.head())
```
