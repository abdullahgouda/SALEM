```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

df = pd.read_csv("../data/processed/reports_clean.csv")

model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

embeddings = model.encode(
    df["description"].tolist(),
    show_progress_bar=True
)
```

```python
new_report = "يوجد حفرة كبيرة أمام البنك الأهلي في شارع عباس العقاد"

new_embedding = model.encode([new_report])

similarities = cosine_similarity(
    new_embedding,
    embeddings
)[0]

best_idx = similarities.argmax()

print("Best Score:", similarities[best_idx])
print("Matched:", df.iloc[best_idx]["description"])
print("Index:", best_idx)
```
